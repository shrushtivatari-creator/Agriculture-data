```python
crop_type=data["Crop_Type"].sum()
soil_type=data["Soil_Type"].sum()
season=data["Season"].nunique()
```

```python
top = data.groupby("Season")["Crop_Type"].value_counts().reset_index(name='Count')
fig1 = px.bar(
    top,
    x="Season",
    y="Count",
    color="Crop_Type",
    text="Count",
    title="Distribution of Crop Types per Season",
    color_continuous_scale="Turbo"
)
fig1.update_layout(template="plotly_dark")
```

```python
# Yield vs Farm Area Scatter Plot
fig2 = px.scatter(
    data,
    x="Farm_Area(acres)",
    y="Yield(tons)",
    color="Crop_Type",
    size="Fertilizer_Used(tons)",
    hover_data=["Irrigation_Type", "Soil_Type"],
    title="Yield vs. Farm Area (Bubble Size = Fertilizer Used)",
    template="plotly_dark"
)
fig2.show()
```

```python
fig3=px.pie(
    data,
    values="Farm_Area(acres)",
    names="Soil_Type",
    color="Soil_Type",
    title="Farm Area Distribution",
    hole=0.6
)
fig3.show()
```

```python
daily=data.groupby("Irrigation_Type")["Fertilizer_Used(tons)"].sum().reset_index()
fig4=px.line(
    daily,
    x="Irrigation_Type",
    y="Fertilizer_Used(tons)",
    markers=True,
    title="Fertilizer is used with each irrigation method"
)
fig4.update_layout(template="plotly_dark")
```

```python
fig5=px.histogram(
    data,
    x="Farm_Area(acres)",
    color="Crop_Type",
    nbins=40,
    color_discrete_sequence=["#fc1403"]
)
```

```python
fig6=px.box(
    data,
    y="Yield(tons)",
    color_discrete_sequence=["#03b5fc"]
)
```

```python
footer_html_content="""
<hr>

<h3 style='text-align:center;
background:#a8329b;
color:#8feb34;
padding:15px;
border-radius:20px;'>
Dashboard Created using
Pandas+Plotly+IPython.display
</h3>


"""
```

```python
fig_dashboard = make_subplots(
    rows=3, cols=2,
    subplot_titles=[
        "Crop Type Distribution",
        "Yield vs. Farm Area (Bubble Size = Fertilizer Used)",
        "Farm Area Distribution",
        "Fertilizer is used with each irrigation method",
        "Fertilizer Usage Distribution",
        "Yield Distribution"

    ],
    specs=[
        [{"type":"xy"},{"type":"xy"}],    # fig1 (bar), fig2 (scatter)
        [{"type":"domain"},{"type":"xy"}], # fig3 (pie), fig4 (line)
        [{"type":"xy"},{"type":"xy"}]      # fig5 (histogram), fig6 (box)
    ]
)
```

```python
for trace in fig1.data:
  fig_dashboard.add_trace(trace,row=1,col=1)

for trace in fig2.data:
  fig_dashboard.add_trace(trace,row=1,col=2)

for trace in fig3.data:
  fig_dashboard.add_trace(trace,row=2,col=1)

for trace in fig4.data:
  fig_dashboard.add_trace(trace,row=2,col=2)

for trace in fig5.data:
  fig_dashboard.add_trace(trace,row=3,col=1)

for trace in fig6.data:
  fig_dashboard.add_trace(trace,row=3,col=2)
```

```python
fig_dashboard.update_layout(
    title_text="Agriculture dataset Dashboard",
    height=1100,
    template="plotly_dark",
    showlegend=False,
)
```

```python
display(HTML(header_html_content))
fig_dashboard.show()
display(HTML(footer_html_content))
```
